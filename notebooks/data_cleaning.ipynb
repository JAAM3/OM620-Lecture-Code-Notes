### Data Cleaning: Removing Duplicats

- Using the fuction `duplicated` and `drop_duplicates`, we can find and handle duplicated rows

```python
data = pd.DataFrame({"k1": ["one", "two"] * 3 + ["two"],
                     "k2": [1,1,2,3,3,4,4]})
print(data)
```

```python
data.duplicated() # what?
```

```python
data.drop_duplicates()
```

- By default, both methods consider all columns. Alternateivly, we can sepcify any subset of columnss to detect or drop duplicates

```python
data.drop_duplicates(subset=["k1"])
```

- both methods also keep the first observed value combination by default. Passing `keep="last"` will return the last one from list

```python
data.drop_duplicates(["k1", "k2"], keep="last")
```

### Data Cleaning: Detecting and filtering outliers

Consider a DataFrame with some normally distrubuted data.

- look into standard deviations

```python
import pandas as pd
import numpy as np
data = pd.DataFrame(np.random.standard_normal ((1000,4))) 
data.describe()
```

Suppose we wanted to find values in one column excedding 3 standard deviations (absolute)

```python
col=data[2] # from the data frame named data, select column 2 and assing it to the variable `col`
# selects the column labeled 2 (3rd one) - the result is a series - not a dataframe

col[col.abs() > 3] # show me the condition of whatever i put in [] - true or false for the absolute value fuction
# show me the original values in column 2 whose absolute values are greater than 3
# .abs() takes the absolute value of each number
# >3 checks whether that abs value exceedes3 which will result in true or false boolean vlaues table
# the outer col[] means select values from the original series
# the innter col.abs()>3 means but only select the positions where this condition is true
```

- the above are technically outliers of the 3 which is the normal range

- to select all rows haveing a value exceeding 3 or -3, you can use the any methond on a Boolean DataFrame

```python
data[(data.abs() >3).any(axis='columns')]
```
