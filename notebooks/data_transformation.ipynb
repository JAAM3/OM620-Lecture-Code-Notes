### Introduction to Data Transformation

- It is very rate that we get the data in exactly the right form we need

- Data transformation includes (usually one of these 5)

- Filter rows by Column attribute

- filtering all the flights that went to LAX

```python
sd22[sd22["dest"] == "LAX"]
```

- Filter rows by `.loc`

- this is the older version of what is above

```python
sd22.loc[(sd22["dest"] =="LAX")]
```

- Filter rows by `.query()`

- this is preferred by data scientits. Notice are all are the same

```python
sd22.query('dest == "LAX"')
```

- Filters can also be combined with logical operatiors

- sd22[(sd22["dest"] == "LAX") & (sd22["dep_delay"]>= 60)] looking at all the flights that went to lax that delayed by more than hour?

```python
sd22[(sd22["dest"] == "LAX") & (sd22["dep_delay"]>= 60)] # 

sd22[(sd22["dest"] == "LAX") & (sd22["dep_delay"]>= 60)].head()

```

- The is the new way of doing the same as above, just using the query function which is better, more descriptive

```python
sd22.query('dest == "LAX" and dep_delay >=60').head()
```

- Other most used operators are | (or), and ~ (not)

- if you use query, you can say 'or' or 'not'

### Filtering: isin

- sometimes we have repeating conditions

- consider the task of filtering Q1 flights

```python
sd22[(sd22["month"] == 1) | (sd22["month"] == 2) | (sd22["month"] == 3)] # this code looks at flights that occur in q1 , have to use 'or' here
```

- The isin fuction allow us to perform such criteria in a compact expression

```python
sd22[sd22["month"].isin([1,2,3])] # by default won't work on and
```

- A similar approach is using the `in` expression with teh `query` function

```python
sd22.query('month in [1,2,3]')
```

### Let's Practice

- Had an arrival delay of two or more hours

```python
sd22.query('arr_delay>=120').sort_values("arr_delay") # probably good idea to look at this way to sort. Order from loweest to highest
```

```python
bay_area_airports = ["SPO", "OAK", "PAO", "SJC"] # should define the objects
sd22.query('dest in @bay_area_airports') # @ tells quary fuction i have defined "bay area" somewhere outside of this.
```

- Departed in Fall

```python
targeted_months=[10,11,12] # if you want to define an object, do it OUTSIDE of the query function. Define it first
sd22.query('month in @targeted_months')
```

- arrived more than two hours late, but didn't leave late

```python
sd22.query('arr_delay> 120 and dep_delay<=0')
```

- Were delayed by at least an hour, but made up over 30 min in flight

```python
sd22.query('dep_delay>=60 and dep_delay - arr_delay>30') # think about what makes the math work
```

- departed between midnight and 6 am (inclusive)

```python
sd22.query('dep_time<=600') # this still works even though the time is in military format. it works here but it may. not work everywhere
```

### Adding a new Column(s)

- We can also expand our data with new features based on the existing columns (or an entirely new external feature)

```python
sd22['total_delay']=sd22['arr_delay'] + sd22['dep_delay']
sd22
```

- By default the new column is palced at the end of the DataFrame. We can select the column position using the insert function.

```python
sd22=sd22.drop(['total_delay'],axis=1)
sd22.insert(0,'total_delay', sd22['arr_delay']+sd22['dep_delay'])
sd22
```

### Adding a new Row(s)

- to extend a DataFrame with a new row, we can utliize the `len` function

```python
data={"state":["Ohio","Ohio","Ohio","Nevada", "Nevada", "Nevada"], 
"year": [2000,2001,2002,2001,2002,2003],
"pop":[1.5,1.7,3.6,2.4,2.9,3.2]}

df=pd.DataFrame(data)

df.loc[len(df)]=["Nevada", 2000, 2.0]

df
```

- A more reliable alternative that works with the multiple rows is to utlizie the concatenation function `concat`

- `Concat` is a better way to do it

```python
new_df=pd.DataFrame({"state":["Nevada"], "year":[2000], "pop":[2.0]})

pd.concat([df, new_df])
```

```python
new_df=pd.DataFrame({"state":["Nevada"], "year":[2000], "pop":[2.0]})
```

```python
new_df=pd.DataFrame({"state":["Nevada"], "year":[2000], "pop":[2.0]})
pd.concat([df, new_df], ignore_index=True)
```
